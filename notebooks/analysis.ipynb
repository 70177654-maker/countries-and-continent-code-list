```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/Countries-Continents.csv')
df.head(10)
```

```python
print('Shape:', df.shape)
print('\nColumns:', df.columns.tolist())
print('\nMissing values:\n', df.isnull().sum())
print('\nContinent value counts:\n', df['Continent'].value_counts())
```

```python
df['NameLength'] = df['Country'].str.len()
print(df.groupby('Continent')['NameLength'].describe())
```

```python
fig, ax = plt.subplots(figsize=(8,4))
df['Continent'].value_counts().plot(kind='bar', ax=ax, color='teal')
ax.set_title('Countries per Continent')
ax.set_ylabel('Count')
plt.tight_layout()
plt.show()
```
